```python
# MitoCarta3.0 advanced analysis
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import networkx as nx

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Optional UMAP
try:
    import umap.umap_ as umap
    HAS_UMAP = True
except Exception:
    HAS_UMAP = False

import plotly.graph_objects as go

sns.set_context("talk")
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (9, 6)


# helpers
def normalize_name(name: str) -> str:
    s = str(name).strip().lower()
    # remove dots, parentheses, brackets, slashes, plus signs
    s = re.sub(r"[.\(\)\[\]{}\/+]", " ", s)
    s = s.replace("-", " ").replace("_", " ")
    s = re.sub(r"\s+", " ", s).strip()
    return s

def normalized_columns_map(df: pd.DataFrame) -> dict:
    """map normalized name to original column name, keep first occurrence"""
    norm_to_orig = {}
    for c in df.columns:
        n = normalize_name(c)
        if n not in norm_to_orig:
            norm_to_orig[n] = c
    return norm_to_orig

def pick_col_fuzzy(df: pd.DataFrame, candidates, required=True, print_hits=True) -> str:
    """
    pick column whose normalized name contains any candidate token,
    tries exact first, then contains, then word boundary contains
    """
    norm_to_orig = normalized_columns_map(df)
    norms = list(norm_to_orig.keys())

    # exact
    for cand in candidates:
        if cand in norms:
            if print_hits:
                print(f"picked exact column for {cand}: {norm_to_orig[cand]}")
            return norm_to_orig[cand]

    # contains
    for cand in candidates:
        for n in norms:
            if cand in n:
                if print_hits:
                    print(f"picked contains column for {cand}: {norm_to_orig[n]}")
                return norm_to_orig[n]

    # word boundary contains
    for cand in candidates:
        pat = re.compile(rf"\b{re.escape(cand)}\b")
        for n in norms:
            if pat.search(n):
                if print_hits:
                    print(f"picked word boundary column for {cand}: {norm_to_orig[n]}")
                return norm_to_orig[n]

    if required:
        raise KeyError(f"could not find any of {candidates} in columns, available normalized names, {norms}")
    return None

def split_multi_values(series):
    def _split(x):
        if pd.isna(x):
            return []
        if isinstance(x, (list, tuple)):
            return list(x)
        s = str(x).strip()
        if not s:
            return []
        for sep in [";", ",", "|", "/"]:
            if sep in s:
                return [t.strip() for t in s.split(sep) if t.strip()]
        return [s]
    return series.apply(_split)

def try_read_excel(path):
    xls = pd.ExcelFile(path)
    # prefer first sheet with many columns
    best = None
    best_cols = 0
    for sh in xls.sheet_names:
        df = pd.read_excel(path, sheet_name=sh, header=0)
        if df.shape[1] > best_cols:
            best = df
            best_cols = df.shape[1]
    return best


# 1, load data
PATH = "Human.MitoCarta3.0.xls"
mito = try_read_excel(PATH)
print("raw columns,")
print(mito.columns.tolist())


# 2, resolve key columns against headers
# we will search using normalized tokens
# file shows, mitocarta3.0 submitolocalization, mitocarta3.0 mitopathways
symbol_col = pick_col_fuzzy(
    mito,
    candidates=[
        "symbol", "gene symbol", "gene", "hgnc symbol"
    ],
    required=True
)

compartment_col = pick_col_fuzzy(
    mito,
    candidates=[
        "submitolocalization", "sub mito localization", "sub mitochondrial localization",
        "mitocarta3.0 submitolocalization", "mitochondrial compartment"
    ],
    required=True
)

pathway_col = pick_col_fuzzy(
    mito,
    candidates=[
        "mitopathways", "mito pathways", "mitocarta3.0 mitopathways", "mito pathway", "pathway"
    ],
    required=True
)

# tissue like numeric columns in sheet look like
# cerebrum total peak intensity log10, heart total peak intensity log10, etc
norm_map = {normalize_name(c): c for c in mito.columns}
tissue_like = []
for c in mito.columns:
    n = normalize_name(c)
    if any(tag in n for tag in [
        "total peak intensity", "log10", "tissue", "abundance", "enrichment", "proteome", "rna"
    ]):
        if pd.to_numeric(mito[c], errors="coerce").notna().sum() > 0:
            tissue_like.append(c)

print("resolved columns,",
      {"symbol": symbol_col, "compartment": compartment_col, "pathway": pathway_col})
print(f"detected numeric tissue like columns, {len(tissue_like)}")

# 3, tidy long format, one compartment and one pathway per gene row
base = pd.DataFrame({
    "Symbol": mito[symbol_col].astype(str),
    "SubMitoLocalization_raw": mito[compartment_col].astype(str),
    "MitoPathway_raw": mito[pathway_col].astype(str)
})

base["SubMitoLocalization_list"] = split_multi_values(base["SubMitoLocalization_raw"])
base["MitoPathway_list"] = split_multi_values(base["MitoPathway_raw"])

df_comp = base.explode("SubMitoLocalization_list").rename(
    columns={"SubMitoLocalization_list": "SubMitoLocalization"}
)
df_comp["SubMitoLocalization"] = df_comp["SubMitoLocalization"].replace("", np.nan).fillna("Unknown")

df_path = base.explode("MitoPathway_list").rename(
    columns={"MitoPathway_list": "MitoPathway"}
)
df_path["MitoPathway"] = df_path["MitoPathway"].replace("", np.nan).fillna("Unknown")

df = df_comp.merge(df_path[["Symbol", "MitoPathway"]], on="Symbol", how="left").drop_duplicates(
    subset=["Symbol", "SubMitoLocalization", "MitoPathway"]
)

print("rows after explode and merge,", df.shape[0])
display(df.head())

# 4, summaries and figures
compartment_counts = df["SubMitoLocalization"].value_counts()

fig, ax = plt.subplots()
sns.barplot(y=compartment_counts.index, x=compartment_counts.values, palette="viridis", ax=ax)
ax.set_title("Distribution of MitoCarta genes across sub mitochondrial compartments")
ax.set_xlabel("Gene count")
ax.set_ylabel("Sub mitochondrial compartment")
plt.tight_layout()
plt.savefig("fig_compartment_distribution.svg")
plt.savefig("fig_compartment_distribution.pdf")
plt.show()

# top pathways
pathway_counts = df["MitoPathway"].value_counts()
top_pathways = pathway_counts.head(15)

fig, ax = plt.subplots(figsize=(11, 8))
sns.barplot(y=top_pathways.index, x=top_pathways.values, palette="mako", ax=ax)
ax.set_title("Top 15 mitochondrial pathways by gene count")
ax.set_xlabel("Gene count")
ax.set_ylabel("Pathway")
plt.tight_layout()
plt.savefig("fig_pathway_counts.svg")
plt.savefig("fig_pathway_counts.pdf")
plt.show()

# contingency stats
cont = pd.crosstab(df["MitoPathway"], df["SubMitoLocalization"])
if cont.shape[0] > 1 and cont.shape[1] > 1:
    chi2, p, dof, expected = stats.chi2_contingency(cont)
    print(f"chi squared association, chi2={chi2:.2f}, p={p:.2e}, dof={dof}")
else:
    print("contingency table too small for chi squared")

# hypergeometric for matrix if present
if "Matrix" in cont.columns:
    M = int(cont.values.sum())
    n = int(cont["Matrix"].sum())
    rows = []
    for pw in cont.index:
        N = int(cont.loc[pw].sum())
        k = int(cont.loc[pw, "Matrix"])
        pval = stats.hypergeom.sf(k - 1, M, n, N)
        rows.append((pw, N, k, pval))
    enrich = pd.DataFrame(rows, columns=["Pathway", "Size", "Matrix_count", "pval"])
    enrich["qval"] = enrich["pval"] * len(enrich)
    enrich = enrich.sort_values("pval")
    enrich.to_csv("table_pathway_matrix_enrichment.csv", index=False)
    with open("table_pathway_matrix_enrichment.tex", "w") as f:
        f.write(enrich.head(20).to_latex(index=False))
    print("top pathways enriched for matrix localization,")
    display(enrich.head(10))
else:
    print("Matrix not present in labels, skipped matrix enrichment")

# sankey
pc = df.groupby(["MitoPathway", "SubMitoLocalization"]).size().reset_index(name="count")
pathways = pc["MitoPathway"].unique().tolist()
comps = pc["SubMitoLocalization"].unique().tolist()
labels = pathways + comps
label_to_idx = {lab: i for i, lab in enumerate(labels)}

fig = go.Figure(data=[go.Sankey(
    node=dict(pad=15, thickness=20, line=dict(color="black", width=0.5), label=labels),
    link=dict(
        source=pc["MitoPathway"].map(label_to_idx),
        target=pc["SubMitoLocalization"].map(label_to_idx),
        value=pc["count"]
    )
)])
fig.update_layout(title_text="Pathway to sub mitochondrial compartment mapping")
fig.write_html("fig_sankey_pathway_compartment.html")
fig.show()

# 5, tissue profiles, PCA and UMAP when available
if len(tissue_like) >= 3:
    tissue_df = mito[tissue_like].apply(pd.to_numeric, errors="coerce").fillna(0.0)
    X = StandardScaler().fit_transform(tissue_df.values)
    pca = PCA(n_components=2, random_state=42)
    pcs = pca.fit_transform(X)
    pcs_df = pd.DataFrame(pcs, columns=["PC1", "PC2"])
    pcs_df["Symbol"] = mito[symbol_col].astype(str).values

    dfp = df.merge(pcs_df, on="Symbol", how="left")

    fig, ax = plt.subplots()
    sns.scatterplot(data=dfp.dropna(subset=["PC1", "PC2"]),
                    x="PC1", y="PC2", hue="SubMitoLocalization", alpha=0.7, ax=ax)
    ax.set_title("PCA of tissue specific mitochondrial profiles")
    plt.tight_layout()
    plt.savefig("fig_pca_tissues.svg")
    plt.savefig("fig_pca_tissues.pdf")
    plt.show()

    if HAS_UMAP:
        reducer = umap.UMAP(random_state=42)
        emb = reducer.fit_transform(X)
        emb_df = pd.DataFrame(emb, columns=["UMAP1", "UMAP2"])
        emb_df["Symbol"] = mito[symbol_col].astype(str).values
        dfu = df.merge(emb_df, on="Symbol", how="left")

        fig, ax = plt.subplots()
        sns.scatterplot(data=dfu.dropna(subset=["UMAP1", "UMAP2"]),
                        x="UMAP1", y="UMAP2", hue="SubMitoLocalization", alpha=0.7, ax=ax)
        ax.set_title("UMAP of tissue specific mitochondrial profiles")
        plt.tight_layout()
        plt.savefig("fig_umap_tissues.svg")
        plt.savefig("fig_umap_tissues.pdf")
        plt.show()
    else:
        print("umap not installed, skipped")
else:
    print("no suitable numeric tissue columns detected, skipped PCA and UMAP")

# 6, network analysis
G = nx.Graph()
for _, r in df.iterrows():
    G.add_node(("gene", r["Symbol"]), bipartite=0)
    G.add_node(("pathway", r["MitoPathway"]), bipartite=1)
    G.add_edge(("gene", r["Symbol"]), ("pathway", r["MitoPathway"]))

deg_cent = nx.degree_centrality(G)
path_deg = [(n[1], c) for n, c in deg_cent.items() if isinstance(n, tuple) and n[0] == "pathway"]
path_deg = sorted(path_deg, key=lambda x: -x[1])[:15]
print("top central pathways,")
for name, score in path_deg:
    print(f"{name}, {score:.4f}")

genes_sample = df["Symbol"].unique().tolist()[:50]
paths_sample = df["MitoPathway"].unique().tolist()[:10]
sub_nodes = [("gene", g) for g in genes_sample] + [("pathway", p) for p in paths_sample]
subG = G.subgraph(sub_nodes).copy()

plt.figure(figsize=(12, 10))
pos = nx.spring_layout(subG, seed=42, k=0.3)
node_colors = ["#3182bd" if n[0] == "gene" else "#31a354" for n in subG.nodes()]
nx.draw(subG, pos, node_color=node_colors, node_size=90, with_labels=False, edge_color="#aaaaaa")
for n, (x, y) in pos.items():
    if n[0] == "pathway":
        plt.text(x, y, n[1][:28], fontsize=6)
plt.title("sample gene pathway network")
plt.tight_layout()
plt.savefig("fig_network_sample.svg")
plt.savefig("fig_network_sample.pdf")
plt.show()

# 7, dual localization enrichment by pathway
n_comp = df.groupby("Symbol")["SubMitoLocalization"].nunique()
dual_genes = set(n_comp[n_comp > 1].index)
print(f"genes with more than one compartment label, {len(dual_genes)}")

fig, ax = plt.subplots()
sns.histplot(n_comp, bins=range(1, n_comp.max() + 2), kde=False, ax=ax)
ax.set_title("multiplicity of sub mitochondrial localization per gene")
ax.set_xlabel("number of compartments")
ax.set_ylabel("gene count")
plt.tight_layout()
plt.savefig("fig_dual_localization.svg")
plt.savefig("fig_dual_localization.pdf")
plt.show()

res = []
all_genes = set(df["Symbol"].unique())
for pw, sub in df.groupby("MitoPathway"):
    genes_pw = set(sub["Symbol"].unique())
    is_dual_pw = sum(g in dual_genes for g in genes_pw)
    not_dual_pw = len(genes_pw) - is_dual_pw
    genes_non = all_genes - genes_pw
    is_dual_non = sum(g in dual_genes for g in genes_non)
    not_dual_non = len(genes_non) - is_dual_non
    table = np.array([[is_dual_pw, not_dual_pw],
                      [is_dual_non, not_dual_non]])
    if table.min() >= 0 and table.sum() > 0:
        odds, pval = stats.fisher_exact(table)
        res.append((pw, len(genes_pw), is_dual_pw, odds, pval))

dual_enrich = pd.DataFrame(res, columns=["Pathway", "Genes_in_pathway", "Dual_genes", "Odds_ratio", "pval"])
if not dual_enrich.empty:
    dual_enrich["qval"] = dual_enrich["pval"] * len(dual_enrich)
    dual_enrich = dual_enrich.sort_values("pval")
    dual_enrich.to_csv("table_dual_loc_enrichment.csv", index=False)
    with open("table_dual_loc_enrichment.tex", "w") as f:
        f.write(dual_enrich.head(20).to_latex(index=False))
    print("top pathways enriched for dual localization genes,")
    display(dual_enrich.head(10))
else:
    print("dual enrichment table is empty, nothing to report")

# 8, save processed outputs
df.to_csv("MitoCarta3_processed_long.csv", index=False)
mito.to_csv("MitoCarta3_original_clean.csv", index=False)

print("saved outputs,")
print("  MitoCarta3_processed_long.csv")
print("  MitoCarta3_original_clean.csv")
print("  fig_compartment_distribution.svg, pdf")
print("  fig_pathway_counts.svg, pdf")
print("  fig_sankey_pathway_compartment.html")
print("  fig_pca_tissues.svg, pdf, if tissue columns present")
print("  fig_umap_tissues.svg, pdf, if UMAP available")
print("  fig_network_sample.svg, pdf")
print("  fig_dual_localization.svg, pdf")
print("  table_pathway_matrix_enrichment.csv, tex, if Matrix present")
print("  table_dual_loc_enrichment.csv, tex")
```

```python
import requests

url = "https://personal.broadinstitute.org/scalvo/MitoCarta3.0/Human.MitoCarta3.0.fasta"
local_filename = "Human.MitoCarta3.0.fasta"

def download_file(url, local_path):
    # Stream the download to avoid loading the whole file into memory
    with requests.get(url, stream=True) as r:
        r.raise_for_status()
        with open(local_path, 'wb') as f:
            for chunk in r.iter_content(chunk_size=8192):
                if chunk:  # filter out keep-alive chunks
                    f.write(chunk)

print(f"Downloading {url} to {local_filename} ...")
download_file(url, local_filename)
print("Download complete.")
```

```python
# MitoCarta sequence grammar scout, plain English, minimal edit designer
# Inputs, Human.MitoCarta3.0.fasta
# Outputs, figures and CSVs with candidate import helices and edit suggestions

import math
import re
from collections import defaultdict
from dataclasses import dataclass
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (9, 5)

# helpers, chemistry and scoring
AA_HELIX_PROP = {
    # Chou Fasman helix propensities, normalized
    "A": 1.45, "C": 0.77, "D": 1.01, "E": 1.51, "F": 1.13,
    "G": 0.53, "H": 1.00, "I": 1.08, "K": 1.16, "L": 1.34,
    "M": 1.20, "N": 0.73, "P": 0.34, "Q": 1.11, "R": 0.79,
    "S": 0.79, "T": 0.82, "V": 1.06, "W": 1.14, "Y": 0.61
}

# Kyte Doolittle hydropathy
AA_HYDRO = {
    "A": 1.8, "C": 2.5, "D": -3.5, "E": -3.5, "F": 2.8,
    "G": -0.4, "H": -3.2, "I": 4.5, "K": -3.9, "L": 3.8,
    "M": 1.9, "N": -3.5, "P": -1.6, "Q": -3.5, "R": -4.5,
    "S": -0.8, "T": -0.7, "V": 4.2, "W": -0.9, "Y": -1.3
}

BASIC = set(["K","R","H"])
ACIDIC = set(["D","E"])

def net_charge(seq, pH=7.4):
    # Simple Henderson Hasselbalch for side chains and termini
    pKa = {"Cterm": 3.1, "Nterm": 8.0, "D": 4.1, "E": 4.1, "H": 6.0, "C": 8.3, "Y": 10.1, "K": 10.5, "R": 12.5}
    charge = 0.0
    if seq:
        charge += 1.0 / (1.0 + 10**(pH - pKa["Nterm"]))
        charge -= 1.0 / (1.0 + 10**(pKa["Cterm"] - pH))
    for aa in seq:
        if aa == "D":
            charge -= 1.0 / (1.0 + 10**(pKa["D"] - pH))
        elif aa == "E":
            charge -= 1.0 / (1.0 + 10**(pKa["E"] - pH))
        elif aa == "H":
            charge += 1.0 / (1.0 + 10**(pH - pKa["H"]))
        elif aa == "K":
            charge += 1.0 / (1.0 + 10**(pH - pKa["K"]))
        elif aa == "R":
            charge += 1.0 / (1.0 + 10**(pH - pKa["R"]))
    return charge

def helix_propensity(seq):
    vals = [AA_HELIX_PROP.get(a, 0.8) for a in seq]
    return float(np.mean(vals)) if vals else 0.0

def hydrophobic_moment(seq, theta=100.0):
    # Eisenberg style moment, theta in degrees per residue for alpha helix
    rad = math.radians(theta)
    vec_x, vec_y = 0.0, 0.0
    for i, aa in enumerate(seq):
        h = AA_HYDRO.get(aa, 0.0)
        ang = i * rad
        vec_x += h * math.cos(ang)
        vec_y += h * math.sin(ang)
    return math.sqrt(vec_x**2 + vec_y**2) / max(len(seq), 1)

def has_helix_breakers(seq):
    # Pro is a breaker, Gly is flexible, strong penalty in long helices
    return ("P" in seq) or (seq.count("G") >= max(1, len(seq)//6))

def window_scores(s, win=18, step=1, nterm_only=True, nterm_limit=120):
    L = len(s)
    end = min(L, nterm_limit) if nterm_only else L
    out = []
    for i in range(0, max(0, end - win + 1), step):
        w = s[i:i+win]
        muH = hydrophobic_moment(w)
        chg = net_charge(w)
        hlx = helix_propensity(w)
        brk = has_helix_breakers(w)
        score = 0.0
        # import like, amphipathic moment high, positive charge, good helix, few breakers
        score += 2.5 * muH
        score += 0.5 * max(0.0, chg)            # positive helps
        score += 1.0 * hlx
        if brk:
            score -= 1.0
        # penalize many acidic residues
        acid_pen = sum(aa in ACIDIC for aa in w) / len(w)
        score -= 1.0 * acid_pen
        out.append((i, i+win, score, muH, chg, hlx, int(brk)))
    cols = ["start","end","score","muH","charge","helix_prop","has_breaker"]
    return pd.DataFrame(out, columns=cols)

# fasta reader
def read_fasta(path):
    records = []
    with open(path, "r") as f:
        seq_id = None
        seq_chunks = []
        for line in f:
            line = line.strip()
            if not line:
                continue
            if line.startswith(">"):
                if seq_id is not None:
                    records.append((seq_id, "".join(seq_chunks).upper()))
                seq_id = line[1:].split()[0]
                seq_chunks = []
            else:
                seq_chunks.append(re.sub("[^A-Za-z]", "", line))
        if seq_id is not None:
            records.append((seq_id, "".join(seq_chunks).upper()))
    return records

# load sequences
FASTA = "Human.MitoCarta3.0.fasta"
recs = read_fasta(FASTA)
print(f"loaded sequences, {len(recs)}")

# Try to pull gene symbol from headers, common RefSeq or UniProt formats
def parse_symbol(h):
    # try common tags, gene, GN, or the first token before a bar
    m = re.search(r"\bGN=([A-Za-z0-9\-_.]+)", h)
    if m:
        return m.group(1)
    m = re.search(r"\bgene[:=]([A-Za-z0-9\-_.]+)", h)
    if m:
        return m.group(1)
    # UniProt like, sp, Q9XYZ1, GENE HUMAN
    parts = h.split("|")
    if len(parts) >= 3:
        tail = parts[-1]
        return tail.split()[0]
    return h.split()[0]

seq_df = pd.DataFrame([{"header": h, "symbol": parse_symbol(h), "seq": s} for h, s in recs])
seq_df["len"] = seq_df["seq"].str.len()

# scan N terminus for import like helices
all_windows = []
best_rows = []
for idx, row in seq_df.iterrows():
    wdf = window_scores(row["seq"], win=18, step=1, nterm_only=True, nterm_limit=120)
    if not wdf.empty:
        wdf["idx"] = idx
        all_windows.append(wdf)
        top = wdf.sort_values("score", ascending=False).head(1).iloc[0].to_dict()
        top["idx"] = idx
        best_rows.append(top)

win_df = pd.concat(all_windows, ignore_index=True) if all_windows else pd.DataFrame()
best_df = pd.DataFrame(best_rows)
if not best_df.empty:
    best_df = best_df.merge(seq_df.reset_index().rename(columns={"index":"idx"}), on="idx", how="left")
    best_df["window_seq"] = best_df.apply(lambda r: r["seq"][int(r["start"]):int(r["end"])], axis=1)

print("top import like window per protein, head")
display(best_df[["symbol","start","end","score","muH","charge","helix_prop","has_breaker","window_seq"]].head(10))

# simple classifier like score at protein level
def protein_import_score(seq):
    # take the best window score in first 120 aa
    w = window_scores(seq, win=18, step=1, nterm_only=True, nterm_limit=120)
    return w["score"].max() if not w.empty else -10.0

seq_df["import_score"] = seq_df["seq"].apply(protein_import_score)

fig, ax = plt.subplots()
ax.hist(seq_df["import_score"], bins=40)
ax.set_title("distribution of import like scores across MitoCarta")
ax.set_xlabel("best N terminal window score")
ax.set_ylabel("count")
plt.tight_layout()
plt.savefig("fig_import_score_hist.svg")
plt.savefig("fig_import_score_hist.pdf")
plt.show()

# minimal edit designer
# goal, flip score up or down with one to three edits in the top window
# rules,
#  add K or R on the hydrophilic face to raise charge,
#  add L or I or V on the hydrophobic face to raise moment,
#  remove D or E inside the window to reduce acidity,
#  avoid creating P inside the helix,
HELIX_DEG = 100.0
def helix_angles(n):
    return [(i * math.radians(HELIX_DEG)) % (2*math.pi) for i in range(n)]

def face_mask(seq, face="hydrophobic"):
    # classify each position by angular sector on the helix wheel
    # choose a sector of 120 degrees for faces
    angs = helix_angles(len(seq))
    # reference vector is angle zero
    mask = []
    for i, aa in enumerate(seq):
        ang = angs[i]
        # project hydropathy onto angle
        # positions near 0 radians are one face, near pi radians the opposite face
        val = math.cos(ang)
        mask.append(val)
    # hydrophobic face, val positive, hydrophilic face, val negative
    if face == "hydrophobic":
        return [i for i, v in enumerate(mask) if v >= 0]
    else:
        return [i for i, v in enumerate(mask) if v < 0]

# candidate substitutions in order of preference
HYDRO_UP = ["L","I","V","F","M"]
CHARGE_UP = ["R","K","H"]
ACID_DOWN = {"D":"N", "E":"Q"}

def propose_edits_for_window(win_seq, max_edits=3, direction="up"):
    edits = []
    seq_list = list(win_seq)

    # reduce acidic residues first
    for i, aa in enumerate(seq_list):
        if aa in ACID_DOWN and len(edits) < max_edits and direction == "up":
            edits.append((i, ACID_DOWN[aa]))
            seq_list[i] = ACID_DOWN[aa]

    # raise hydrophobic moment on hydrophobic face
    if direction == "up":
        for i in face_mask(win_seq, face="hydrophobic"):
            if len(edits) >= max_edits:
                break
            aa = seq_list[i]
            if aa in "AGSTCPYWHNQ" and aa != "P":
                seq_list[i] = HYDRO_UP[0]
                edits.append((i, seq_list[i]))

    # raise positive charge on hydrophilic face
    if direction == "up":
        for i in face_mask(win_seq, face="hydrophilic"):
            if len(edits) >= max_edits:
                break
            aa = seq_list[i]
            if aa in "AGSTCPYWHNQ" and aa != "P":
                seq_list[i] = CHARGE_UP[0]
                edits.append((i, seq_list[i]))

    # direction down, weaken import, replace basic with neutral, replace hydrophobics with small polar
    if direction == "down":
        for i, aa in enumerate(seq_list):
            if len(edits) >= max_edits:
                break
            if aa in "KRH":
                seq_list[i] = "N"
                edits.append((i, "N"))
        for i, aa in enumerate(seq_list):
            if len(edits) >= max_edits:
                break
            if aa in "LIVFMWY":
                seq_list[i] = "S"
                edits.append((i, "S"))

    new_seq = "".join(seq_list)
    return edits, new_seq

def apply_edits(full_seq, start, edits):
    lst = list(full_seq)
    for rel_i, new_aa in edits:
        lst[start + rel_i] = new_aa
    return "".join(lst)

def evaluate_window(seq, start, end):
    w = seq[start:end]
    s = window_scores(seq, win=end-start, step=1, nterm_only=False)
    row = s[s["start"]==start].iloc[0]
    return dict(score=row["score"], muH=row["muH"], charge=row["charge"], helix=row["helix_prop"], breaker=bool(row["has_breaker"]))

# build proposals for each protein
proposals = []
for _, r in best_df.iterrows():
    sym = r["symbol"]
    seq = r["seq"]
    s0 = int(r["start"]); e0 = int(r["end"])
    win = seq[s0:e0]

    base = evaluate_window(seq, s0, e0)

    # up edits
    up_edits, up_win = propose_edits_for_window(win, max_edits=3, direction="up")
    seq_up = apply_edits(seq, s0, up_edits)
    up = evaluate_window(seq_up, s0, e0)

    # down edits
    dn_edits, dn_win = propose_edits_for_window(win, max_edits=3, direction="down")
    seq_dn = apply_edits(seq, s0, dn_edits)
    dn = evaluate_window(seq_dn, s0, e0)

    proposals.append({
        "symbol": sym,
        "start": s0,
        "end": e0,
        "orig_window": win,
        "orig_score": base["score"],
        "orig_muH": base["muH"],
        "orig_charge": base["charge"],
        "orig_helix": base["helix"],
        "up_edits": ";".join([f"{win[i]}{i+1}{aa}" for i, aa in up_edits]),
        "up_window": up_win,
        "up_score": up["score"],
        "up_muH": up["muH"],
        "up_charge": up["charge"],
        "up_helix": up["helix"],
        "down_edits": ";".join([f"{win[i]}{i+1}{aa}" for i, aa in dn_edits]),
        "down_window": dn_win,
        "down_score": dn["score"],
        "down_muH": dn["muH"],
        "down_charge": dn["charge"],
        "down_helix": dn["helix"],
    })

prop_df = pd.DataFrame(proposals)
prop_df["delta_up"] = prop_df["up_score"] - prop_df["orig_score"]
prop_df["delta_down"] = prop_df["down_score"] - prop_df["orig_score"]

# rank strongest predicted flips
prop_top_activate = prop_df.sort_values("delta_up", ascending=False).head(25)
prop_top_deactivate = prop_df.sort_values("delta_down").head(25)

print("top candidates to increase import like signal, head")
display(prop_top_activate[["symbol","start","end","orig_score","up_score","delta_up","up_edits","orig_window","up_window"]].head(10))

print("top candidates to reduce import like signal, head")
display(prop_top_deactivate[["symbol","start","end","orig_score","down_score","delta_down","down_edits","orig_window","down_window"]].head(10))

# save tables
seq_df.sort_values("import_score", ascending=False).to_csv("table_import_scores.csv", index=False)
prop_df.to_csv("table_edit_proposals_all.csv", index=False)
prop_top_activate.to_csv("table_edit_proposals_activate_top25.csv", index=False)
prop_top_deactivate.to_csv("table_edit_proposals_deactivate_top25.csv", index=False)

# per residue maps for a few examples
def plot_window_profile(symbol, seq, start, end, title_prefix="original"):
    w = seq[start:end]
    xs = list(range(start+1, end+1))
    mu = []
    ch = []
    hp = []
    br = []
    # per residue contributions, approximate by single point changes
    base_mu = hydrophobic_moment(w)
    base_ch = net_charge(w)
    base_hp = helix_propensity(w)
    for i, aa in enumerate(w):
        # contribution proxies
        mu.append(AA_HYDRO.get(aa, 0.0))
        ch.append(1 if aa in BASIC else (-1 if aa in ACIDIC else 0))
        hp.append(AA_HELIX_PROP.get(aa, 0.8))
        br.append(1 if aa == "P" else 0)

    fig, axes = plt.subplots(3, 1, figsize=(10,7), sharex=True)
    axes[0].plot(xs, mu)
    axes[0].set_ylabel("hydropathy proxy")
    axes[1].plot(xs, ch)
    axes[1].set_ylabel("charge proxy")
    axes[2].plot(xs, hp)
    axes[2].set_ylabel("helix prop")
    axes[2].set_xlabel("position")
    fig.suptitle(f"{title_prefix} window for {symbol}, start {start+1}, end {end}, mu {base_mu:.2f}, charge {base_ch:.2f}, helix {base_hp:.2f}")
    plt.tight_layout()
    plt.savefig(f"fig_{symbol}_{title_prefix}_window_profile.svg")
    plt.savefig(f"fig_{symbol}_{title_prefix}_window_profile.pdf")
    plt.show()

# plot for top three activations
for _, r in prop_top_activate.head(3).iterrows():
    plot_window_profile(r["symbol"], seq_df.loc[seq_df["symbol"]==r["symbol"],"seq"].values[0], int(r["start"]), int(r["end"]), title_prefix="original")
    plot_window_profile(r["symbol"], apply_edits(seq_df.loc[seq_df["symbol"]==r["symbol"],"seq"].values[0], int(r["start"]),
                                                 [(int(ed[1:-1])-1, ed[-1]) for ed in r["up_edits"].split(";") if ed]),
                        int(r["start"]), int(r["end"]), title_prefix="up")

# summary figure
fig, ax = plt.subplots()
ax.scatter(prop_df["orig_score"], prop_df["up_score"], s=10, alpha=0.5, label="up edits")
ax.scatter(prop_df["orig_score"], prop_df["down_score"], s=10, alpha=0.5, label="down edits")
ax.plot([prop_df[["orig_score","up_score","down_score"]].min().min(), prop_df[["orig_score","up_score","down_score"]].max().max()],
        [prop_df[["orig_score","up_score","down_score"]].min().min(), prop_df[["orig_score","up_score","down_score"]].max().max()],
        lw=1)
ax.set_xlabel("original best window score")
ax.set_ylabel("edited best window score")
ax.set_title("effect of minimal edits on import like score")
ax.legend()
plt.tight_layout()
plt.savefig("fig_edit_effects_scatter.svg")
plt.savefig("fig_edit_effects_scatter.pdf")
plt.show()

print("saved,")
print("  table_import_scores.csv")
print("  table_edit_proposals_all.csv")
print("  table_edit_proposals_activate_top25.csv")
print("  table_edit_proposals_deactivate_top25.csv")
print("  fig_import_score_hist.svg and pdf")
print("  fig_edit_effects_scatter.svg and pdf")
print("  per window profile figures for top examples")
```

```python
# Cell 2, read FASTA and prepare batches
import re
import numpy as np
import pandas as pd

def read_fasta(path):
    recs = []
    with open(path) as f:
        header = None
        seq = []
        for line in f:
            line = line.strip()
            if not line:
                continue
            if line.startswith(">"):
                if header:
                    recs.append((header, "".join(seq)))
                header = line[1:]
                seq = []
            else:
                seq.append(line)
        if header:
            recs.append((header, "".join(seq)))
    return pd.DataFrame(recs, columns=["header","seq"])

df = read_fasta("Human.MitoCarta3.0.fasta")
print("Sequences loaded,", len(df))

# optional, trim ultra long sequences to a max length for speed
MAX_LEN = 1024
df["seq_trim"] = df["seq"].str.slice(0, MAX_LEN)
```

```python
# Cell 3, load ESM two tiny model and compute per sequence embeddings
import torch
import esm
from tqdm.auto import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model, alphabet = esm.pretrained.esm2_t6_8M_UR50D()  # tiny model, fast on CPU
model = model.to(device)
model.eval()
batch_converter = alphabet.get_batch_converter()

def embed_batch(ids, seqs):
    batch = [(i, s) for i, s in zip(ids, seqs)]
    labels, strs, toks = batch_converter(batch)
    toks = toks.to(device)
    with torch.no_grad():
        out = model(toks, repr_layers=[6])
    reps = out["representations"][6]  # shape, batch, L, H
    # mean pool over residues excluding special tokens
    emb = []
    for i, s in enumerate(strs):
        L = len(s)
        # tokens include start and end, take positions 1 to L
        em = reps[i, 1:L+1, :].mean(0).cpu().numpy()
        emb.append(em)
    return np.stack(emb)

# batch over sequences
BATCH = 64
emb_list = []
for i in tqdm(range(0, len(df), BATCH)):
    sub = df.iloc[i:i+BATCH]
    ids = sub["header"].tolist()
    seqs = sub["seq_trim"].tolist()
    emb = embed_batch(ids, seqs)
    emb_list.append(emb)

X = np.vstack(emb_list)
print("Embedding matrix shape,", X.shape)

# save to disk
np.save("mitocarta_esm2_embeddings.npy", X)
df[["header","seq_trim"]].to_csv("mitocarta_esm2_meta.csv", index=False)
print("Saved mitocarta_esm2_embeddings.npy and mitocarta_esm2_meta.csv")
```

```python
def normalize_name(name):
    s = str(name).strip().lower()
    s = re.sub(r"[.\(\)\[\]{}\/+]", " ", s)
    s = s.replace("-", " ").replace("_", " ")
    s = re.sub(r"\s+", " ", s).strip()
    return s

def split_multi_values(series):
    def _split(x):
        if pd.isna(x): return []
        s = str(x).strip()
        if not s: return []
        for sep in [";", ",", "|", "/"]:
            if sep in s:
                return [t.strip() for t in s.split(sep) if t.strip()]
        return [s]
    return series.apply(_split)

# pick the first sheet that looks like the main table
xls = pd.ExcelFile("Human.MitoCarta3.0.xls")
best = None
best_cols = 0
for sh in xls.sheet_names:
    df_try = pd.read_excel("Human.MitoCarta3.0.xls", sheet_name=sh)
    if df_try.shape[1] > best_cols and df_try.shape[0] > 100:
        best, best_cols = df_try, df_try.shape[1]

if best is None:
    raise RuntimeError("could not find a suitable sheet in the Excel")

mito_raw = best.copy()
mito_norm_cols = {c: normalize_name(c) for c in mito_raw.columns}
mito = mito_raw.rename(columns=mito_norm_cols)

# find columns by fuzzy tokens
def pick_col(df, tokens):
    names = list(df.columns)
    for t in tokens:
        for c in names:
            if t in c:
                return c
    return None

col_symbol = pick_col(mito, ["symbol", "gene symbol", "gene"])
col_comp   = pick_col(mito, ["sub mito localization", "submitolocalization", "sub mitochondrial localization"])
col_path   = pick_col(mito, ["mitopathways", "mito pathway", "mitopathway", "pathway"])

print("using columns,", {"symbol": col_symbol, "compartment": col_comp, "pathway": col_path})

# build long tidy table with one compartment and one pathway per row
base = pd.DataFrame({
    "Symbol": mito[col_symbol].astype(str),
    "SubMitoLocalization_raw": mito[col_comp].astype(str),
    "MitoPathway_raw": mito[col_path].astype(str)
})
df_comp = base.assign(SubMitoLocalization_list=split_multi_values(base["SubMitoLocalization_raw"])) \
              .explode("SubMitoLocalization_list") \
              .rename(columns={"SubMitoLocalization_list": "SubMitoLocalization"})
df_comp["SubMitoLocalization"] = df_comp["SubMitoLocalization"].replace("", np.nan).fillna("Unknown")

df_path = base.assign(MitoPathway_list=split_multi_values(base["MitoPathway_raw"])) \
              .explode("MitoPathway_list") \
              .rename(columns={"MitoPathway_list": "MitoPathway"})
df_path["MitoPathway"] = df_path["MitoPathway"].replace("", np.nan).fillna("Unknown")

df_long = df_comp.merge(df_path[["Symbol", "MitoPathway"]], on="Symbol", how="left").drop_duplicates(
    subset=["Symbol", "SubMitoLocalization", "MitoPathway"]
)
df_long.to_csv("MitoCarta3_processed_long.csv", index=False)
print("long tidy saved, shape", df_long.shape)
```
